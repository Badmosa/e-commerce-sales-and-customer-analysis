# Analysis
## 1. SALES & REVENUE

```python
# Revenue and order volume by sales channel.
sales_by_channel =  df.groupby("sales_channel").agg(orders=("order_id", "count"),
             gross_sales=("gross_sales", "sum"),
             net_sales=("net_sales", "sum")).sort_values("net_sales", ascending=False)


sales_by_channel
```

```python
# Revenue by region
sales_by_region = df.groupby("region").agg(
    orders=("order_id", "count"), net_sales=("net_sales", "sum")).sort_values(
        "net_sales", ascending=False)

sales_by_region
```

```python
# Net sales trend over time.
# Frequent date.

d = df.copy()
d["order_date"] = pd.to_datetime(d["order_date"])

d.set_index("order_date").resample("ME")["net_sales"].sum()
```

## 2. PROFITABILITY

```python
# Most profitable regions and channels, not just highest revenue.
profit_by_region_channel = df.groupby(["region", "sales_channel"]).agg(
        net_sales=("net_sales", "sum"),
        profit=("profit", "sum"),
        avg_margin_pct=("profit_margin_percentage", "mean")).sort_values("profit", ascending=False)

profit_by_region_channel
```

```python
# Correlation between discount amount and profit margin.
discount_impact_on_margin = df[["discount_amount", "profit_margin_percentage"]].corr().iloc[0, 1]
coupon_vs_no_coupon = (
        df.assign(used_coupon=df["coupon_code"].notna())
        .groupby("used_coupon")["profit_margin_percentage"]
        .mean()
    )

discount_impact_on_margin
```

```python
# Checking where the biggest cost drains are: product cost vs shipping cost.
cost_breakdown = df[["product_cost", "shipping_cost", "tax_amount", "discount_amount"]].sum()

cost_breakdown

cost_breakdown
```

## 3. CUSTOMER BEHAVIOR & SEGMENTATION

```python
# Highest-value customers by customer_lifetime_value.
top_customers_by_ltv = (
    df.drop_duplicates("customer_id")
      .sort_values("customer_lifetime_value", ascending=False)
      [["customer_id", "customer_name", "customer_lifetime_value", "customer_order_count"]]
      .head()
)

top_customers_by_ltv
```

```python
# the rate of returning customers
returning_customer_rate = df["is_repeat_customer"].mean() * 100
spend_comparison = df.groupby("is_repeat_customer")["net_sales"].mean()

returning_customer_rate

# summary of the returning customer rate
summary = pd.DataFrame({
    "Metric": [
        "Returning Customer Rate",
        "Average Net Sales"
    ],
    "Value": [
        returning_customer_rate,
        spend_comparison.mean()
    ]
}).round(2)

summary
```

```python
# Spending and order patterns by customer segment
spend_by_segment = df.groupby(["customer_segment", "customer_type"]).agg(orders=("order_id", "count"),
        avg_net_sales=("net_sales", "mean"),
        avg_ltv=("customer_lifetime_value", "mean")).sort_values("avg_ltv", ascending=False)

spend_by_segment.head()
```

```python
# # Buying patterns by age and gender.

spend_by_demographics = df.copy()
spend_by_demographics["age_bracket"] = pd.cut(
    spend_by_demographics["customer_age"],
    bins=[0, 18, 25, 35, 45, 55, 65, 100],
    labels=["<18", "18-25", "26-35", "36-45", "46-55", "56-65", "65+"]
)

spend_by_demographics = (
    spend_by_demographics
    .groupby(["age_bracket", "gender"], observed=True)
    .agg(
        orders=("order_id", "count"),
        avg_net_sales=("net_sales", "mean")
    )
    .round(2)
    .reset_index()
)

spend_by_demographics
```

## 4. MARKETING EFFECTIVENESS.

```python
# Net sales and order volume by marketing channel and campaign.
channel_conversion_value = df.groupby(["marketing_channel", "campaign_name"]).agg(orders=("order_id", "count"),
            net_sales=("net_sales", "sum"),
             avg_order_value=("net_sales", "mean")).sort_values("net_sales", ascending=False)

channel_conversion_value
```

```python
# Compare order value and margin for coupon vs non-coupon orders.
coupon_incremental_impact = df.assign(used_coupon=df["coupon_code"].notna()).groupby("used_coupon").agg(
        orders=("order_id", "count"),
        avg_net_sales=("net_sales", "mean"),
        avg_profit_margin=("profit_margin_percentage", "mean"),
)

coupon_incremental_impact
```

```python
# percentage of orders where actual delivery exceeded estimated delivery days.
delivery_sla_breach_rate = df.copy()

delivery_sla_breach_rate["sla_breached"] = (
    delivery_sla_breach_rate["delivery_days"]
    > delivery_sla_breach_rate["estimated_delivery_days"]
)

delivery_sla_breach_rate = {
    "overall_breach_rate_pct": round(
        delivery_sla_breach_rate["sla_breached"].mean() * 100, 2
    ),
    "breach_rate_by_warehouse": (
        delivery_sla_breach_rate
        .groupby("warehouse")["sla_breached"]
        .mean()
        .mul(100)
        .round(2)
    ),
    "breach_rate_by_shipping_method": (
        delivery_sla_breach_rate
        .groupby("shipping_method")["sla_breached"]
        .mean()
        .mul(100)
        .round(2)
    )
}

delivery_sla_breach_rate
```
