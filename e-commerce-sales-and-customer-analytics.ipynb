# Analysis
## 1. SALES & REVENUE

```python
# Revenue and order volume by sales channel.
sales_by_channel =  df.groupby("sales_channel").agg(orders=("order_id", "count"),
             gross_sales=("gross_sales", "sum"),
             net_sales=("net_sales", "sum")).sort_values("net_sales", ascending=False)


sales_by_channel
```

```python
# Revenue by region
sales_by_region = df.groupby("region").agg(
    orders=("order_id", "count"), net_sales=("net_sales", "sum")).sort_values(
        "net_sales", ascending=False)

sales_by_region
```

```python
# Net sales trend over time.
# Frequent date.

d = df.copy()
d["order_date"] = pd.to_datetime(d["order_date"])

d.set_index("order_date").resample("ME")["net_sales"].sum()
```

## 2. PROFITABILITY

```python
# Most profitable regions and channels, not just highest revenue.
profit_by_region_channel = df.groupby(["region", "sales_channel"]).agg(
        net_sales=("net_sales", "sum"),
        profit=("profit", "sum"),
        avg_margin_pct=("profit_margin_percentage", "mean")).sort_values("profit", ascending=False)

profit_by_region_channel
```

```python
# Correlation between discount amount and profit margin.
discount_impact_on_margin = df[["discount_amount", "profit_margin_percentage"]].corr().iloc[0, 1]
coupon_vs_no_coupon = (
        df.assign(used_coupon=df["coupon_code"].notna())
        .groupby("used_coupon")["profit_margin_percentage"]
        .mean()
    )

discount_impact_on_margin
```

```python
# Checking where the biggest cost drains are: product cost vs shipping cost.
cost_breakdown = df[["product_cost", "shipping_cost", "tax_amount", "discount_amount"]].sum()

cost_breakdown

cost_breakdown
```

## 3. CUSTOMER BEHAVIOR & SEGMENTATION

```python
# Highest-value customers by customer_lifetime_value.
top_customers_by_ltv = (
    df.drop_duplicates("customer_id")
      .sort_values("customer_lifetime_value", ascending=False)
      [["customer_id", "customer_name", "customer_lifetime_value", "customer_order_count"]]
      .head()
)

top_customers_by_ltv
```

```python
# the rate of returning customers
returning_customer_rate = df["is_repeat_customer"].mean() * 100
spend_comparison = df.groupby("is_repeat_customer")["net_sales"].mean()

returning_customer_rate

# summary of the returning customer rate
summary = pd.DataFrame({
    "Metric": [
        "Returning Customer Rate",
        "Average Net Sales"
    ],
    "Value": [
        returning_customer_rate,
        spend_comparison.mean()
    ]
}).round(2)

summary
```
